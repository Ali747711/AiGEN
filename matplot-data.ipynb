## Bar Chart with Numpy and Matplot

```python
import numpy as np
import matplotlib.pyplot as plt

# Data for the bar chart
categories = np.array(['A', 'B', 'C', 'D'])
values = np.array([15, 24, 12, 30])

# Create the bar chart
plt.bar(categories, values, color='skyblue', edgecolor='black')

# Add title and labels
plt.title("Bar Chart Tutorial")
plt.xlabel("Categories")
plt.ylabel("Values")

# Show the plot
plt.show()
```

## Pie Chart with Numpy and Matplot

```python
import numpy as np
import matplotlib.pyplot as plt

# Data for the pie chart using NumPy arrays
labels = np.array(['Python', 'Java', 'C++', 'JavaScript'])
sizes = np.array([45, 25, 15, 15])
explode = np.array([0.1, 0, 0, 0])  # Explode the 1st slice (Python)

# Create the pie chart
plt.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%',
        shadow=True, startangle=140, colors=['gold', 'yellowgreen', 'lightcoral', 'lightskyblue'])

# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')
plt.title("Programming Languages Popularity (Pie Chart)")

# Show plot
plt.show()
```

## Comprehensive Subplots Grid Tutorial

```python
# Configure a 3x2 grid of subplots
fig, axs = plt.subplots(3, 2, figsize=(15, 18))
fig.suptitle('All Tutorial Charts Together', fontsize=20, fontweight='bold', y=0.95)

# 1. Line Plot (top-left)
x1 = np.array([1, 2, 3, 4])
y1 = np.array([10, 20, 35, 40])
y2 = np.array([3, 8, 1, 10])
axs[0, 0].plot(x1, y1, label='Line 1', color='red', marker='o')
axs[0, 0].plot(x1, y2, label='Line 2', color='green', marker='o', linestyle='--')
axs[0, 0].set_title('Line Plot')
axs[0, 0].set_xlabel('x-axis')
axs[0, 0].set_ylabel('y-axis')
axs[0, 0].legend()

# 2. Bar Chart (top-right)
categories = np.array(['A', 'B', 'C', 'D'])
values = np.array([15, 24, 12, 30])
axs[0, 1].bar(categories, values, color='skyblue', edgecolor='black')
axs[0, 1].set_title('Bar Chart')
axs[0, 1].set_xlabel('Categories')
axs[0, 1].set_ylabel('Values')

# 3. Scatter Plot (middle-left)
x_scat = np.array([1, 2, 3, 4, 5])
y_scat = np.array([5, 4, 6, 3, 7])
axs[1, 0].scatter(x_scat, y_scat, color='purple', s=150, alpha=0.7, edgecolor='black')
axs[1, 0].set_title('Scatter Plot')
axs[1, 0].set_xlabel('X Value')
axs[1, 0].set_ylabel('Y Value')

# 4. Histogram (middle-right)
np.random.seed(42)
data_hist = np.random.randn(1000)
axs[1, 1].hist(data_hist, bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
axs[1, 1].set_title('Histogram')
axs[1, 1].set_xlabel('Value')
axs[1, 1].set_ylabel('Frequency')

# 5. Pie Chart (bottom-left)
labels = np.array(['Python', 'Java', 'C++', 'JavaScript'])
sizes = np.array([45, 25, 15, 15])
explode = np.array([0.1, 0, 0, 0])
axs[2, 0].pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%',
              shadow=True, startangle=140, colors=['gold', 'yellowgreen', 'lightcoral', 'lightskyblue'])
axs[2, 0].set_title('Pie Chart')

# 6. Box Plot (bottom-right)
np.random.seed(42)
box_data1 = np.random.normal(100, 10, 200)
box_data2 = np.random.normal(90, 20, 200)
axs[2, 1].boxplot([box_data1, box_data2], tick_labels=['Group 1', 'Group 2'], patch_artist=True,
                  boxprops=dict(facecolor='lightblue', color='blue'))
axs[2, 1].set_title('Box Plot')
axs[2, 1].set_ylabel('Values')

# Adjust spacing and display layout
plt.tight_layout(rect=[0, 0.03, 1, 0.93])
plt.show()
```

## Next Steps: Advanced Matplotlib Features
Now that you know how to build basic plots and organize them into subplots, let's explore three powerful, advanced features of Matplotlib:
1. **Twin Axes (Secondary Y-Axis)**: Useful when comparing two metrics with different units.
2. **Advanced Annotations**: Drawing attention to specific data points.
3. **3D Plotting**: Utilizing `mplot3d` to visualize multivariate relations.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Secondary Y-Axis Example (Twin Axes) ---
fig, ax1 = plt.subplots(figsize=(10, 5))

time = np.arange(0, 10, 0.1)
temperature = 20 + 5 * np.sin(time)     # Primary metric (e.g., °C)
humidity = 50 + 20 * np.cos(time)        # Secondary metric (e.g., %)

# Plot temperature on primary axis
color = 'tab:red'
ax1.set_xlabel('Time (hours)', fontweight='bold')
ax1.set_ylabel('Temperature (°C)', color=color, fontweight='bold')
line1 = ax1.plot(time, temperature, color=color, label='Temperature')
ax1.tick_params(axis='y', labelcolor=color)

# Create a secondary y-axis sharing the same x-axis
ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Humidity (%)', color=color, fontweight='bold')
line2 = ax2.plot(time, humidity, color=color, linestyle='--', label='Humidity')
ax2.tick_params(axis='y', labelcolor=color)

# Combine legends from both axes
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Using Twin Y-Axes to Compare Different Metrics', fontsize=14, pad=15)
plt.show()
```

```python
# --- 2. Advanced Annotations Example ---
plt.figure(figsize=(10, 5))
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='Sine Wave', color='purple')

# Add custom annotation highlighting the peak of the wave
plt.annotate('Local Maximum (Peak)',
             xy=(np.pi/2, 1.0),                  # Point to annotate
             xytext=(np.pi/2 + 1, 0.8),          # Text location
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
             fontsize=10,
             bbox=dict(boxstyle="round,pad=0.3", fc="yellow", alpha=0.5))

plt.title('Advanced Annotations & Custom Bounding Boxes', fontsize=14)
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
# --- 3. 3D Plotting Example ---
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Generate meshgrid data
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot 3D Surface
surf = ax.plot_surface(X, Y, Z, cmap='coolwarm', edgecolor='none', alpha=0.85)

# Customize labels and colorbar
ax.set_title('3D Surface Plot using mplot3d', fontsize=14)
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5, label='Z elevation')

plt.show()
```
